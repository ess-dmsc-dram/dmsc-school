```python
def gaussian(x, mu, sigma):
    return 1/(np.sqrt(2*np.pi*sigma**2))*np.exp(-0.5*((x-mu)/sigma)**2)
```

```python
x = np.arange(6.46, 6.54, 0.0005)
mu = 6.5
sigma = 0.01
```

```python
fix, ax = plt.subplots()
ax.fill_between(x,gaussian(x, mu, sigma), alpha=0.7, color='dodgerblue')
ax.set_xlabel('y')
ax.set_ylabel('p(y)')
ax.set_ylim(0.0, 50)
ax.set_xlim(6.46, 6.54)
plt.show()
```

```python
from scipy.special import factorial
def poisson(k, lambda_):
    return (lambda_**k * np.exp(-lambda_)) / factorial(k)
```

```python
x = np.arange(0, 21, 1)
y = poisson(x, 3)
```

```python
fix, ax = plt.subplots()
ax.stem(x,y, basefmt=" ", linefmt='dodgerblue')
ax.set_xlabel('k')
ax.set_ylabel('p(k)')
ax.set_ylim(-0.005, 0.24)
ax.set_xticks(np.arange(0, 21, 1), minor=True)
ax.set_xticks(np.arange(0, 21, 2), minor=False)
```
